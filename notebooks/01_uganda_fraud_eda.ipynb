```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
sns.set_theme(style="whitegrid")

# Load Ugandan transaction dataset
df = pd.read_csv("../data/training.csv")

print(f"Total Ugandan Transactions: {df.shape[0]:,} rows, {df.shape[1]} columns")
fraud_counts = df["FraudResult"].value_counts()
fraud_pct = df["FraudResult"].mean() * 100
print(f"Legitimate (0): {fraud_counts[0]:,} | Fraudulent (1): {fraud_counts[1]:,} ({fraud_pct:.3f}%)")

display(df.head())
```

```python
# Parse UTC/EAT timestamps (Uganda is UTC+3)
df["TransactionStartTime"] = pd.to_datetime(df["TransactionStartTime"])
df["Hour_EAT"] = (df["TransactionStartTime"].dt.hour + 3) % 24
df["DayOfWeek"] = df["TransactionStartTime"].dt.day_name()

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Fraud Rate by Product Category
cat_fraud = df.groupby("ProductCategory")["FraudResult"].mean().sort_values(ascending=False)
sns.barplot(x=cat_fraud.values * 100, y=cat_fraud.index, ax=axes[0, 0], palette="Reds_r")
axes[0, 0].set_title("Fraud Rate (%) by Product Category")
axes[0, 0].set_xlabel("Fraud Rate (%)")

# 2. Fraud Rate by ChannelId
chan_fraud = df.groupby("ChannelId")["FraudResult"].mean().sort_values(ascending=False)
sns.barplot(x=chan_fraud.index, y=chan_fraud.values * 100, ax=axes[0, 1], palette="Blues_d")
axes[0, 1].set_title("Fraud Rate (%) by Channel")
axes[0, 1].set_ylabel("Fraud Rate (%)")

# 3. Fraud Rate by Hour of Day (Kampala / EAT Time)
hourly_fraud = df.groupby("Hour_EAT")["FraudResult"].mean() * 100
axes[1, 0].plot(hourly_fraud.index, hourly_fraud.values, marker="o", color="crimson", linewidth=2)
axes[1, 0].set_title("Fraud Rate (%) by Hour of Day (EAT / Uganda Time)")
axes[1, 0].set_xlabel("Hour of Day (0 - 23 EAT)")
axes[1, 0].set_ylabel("Fraud Rate (%)")
axes[1, 0].set_xticks(range(0, 24, 2))

# 4. Transaction Value Distribution (Log Scale): Legit vs Fraud
sns.boxplot(data=df, x="FraudResult", y=np.log1p(df["Value"]), ax=axes[1, 1], palette="Set2")
axes[1, 1].set_title("Log(Transaction Value in UGX): Legit (0) vs Fraud (1)")
axes[1, 1].set_ylabel("log1p(UGX Value)")

plt.tight_layout()
plt.show()
```

```python
def engineer_fraud_features(data: pd.DataFrame) -> pd.DataFrame:
    df_f = data.copy()
    df_f["TransactionStartTime"] = pd.to_datetime(df_f["TransactionStartTime"])
    
    # Sort chronologically so rolling/expanding calculations never peek into the future
    df_f = df_f.sort_values("TransactionStartTime").reset_index(drop=True)
    
    # 1. Time features (East Africa Time = UTC + 3)
    df_f["Hour_EAT"] = (df_f["TransactionStartTime"].dt.hour + 3) % 24
    df_f["DayOfWeek_Num"] = df_f["TransactionStartTime"].dt.dayofweek
    df_f["Is_Night_EAT"] = df_f["Hour_EAT"].isin([23, 0, 1, 2, 3, 4]).astype(int)
    
    # 2. Amount & UGX Value features
    df_f["Is_Debit"] = (df_f["Amount"] > 0).astype(int)
    df_f["Log_Value"] = np.log1p(df_f["Value"])
    df_f["Fee_Diff"] = np.abs(df_f["Amount"]) - df_f["Value"]
    
    # 3. Customer Velocity & Historical Spend Ratio (past-only expanding window!)
    df_f["Secs_Since_Last_Tx"] = (
        df_f.groupby("CustomerId")["TransactionStartTime"]
        .diff()
        .dt.total_seconds()
        .fillna(999999)
    )
    df_f["Cust_Past_Tx_Count"] = df_f.groupby("CustomerId").cumcount()
    
    # Expanding mean of past transaction values (shifted by 1 so current tx isn't included)
    past_sum = df_f.groupby("CustomerId")["Value"].cumsum() - df_f["Value"]
    df_f["Cust_Past_Mean_Value"] = np.where(
        df_f["Cust_Past_Tx_Count"] > 0,
        past_sum / df_f["Cust_Past_Tx_Count"],
        df_f["Value"]
    )
    df_f["Value_to_Cust_Mean"] = df_f["Value"] / (df_f["Cust_Past_Mean_Value"] + 1)
    
    # 4. Batch size (how many transactions share the same BatchId)
    batch_counts = df_f["BatchId"].value_counts()
    df_f["Batch_Size"] = df_f["BatchId"].map(batch_counts)
    
    # 5. Drop raw high-cardinality IDs & constant columns (CurrencyCode/CountryCode are all UGX/256)
    drop_cols = [
        "TransactionId", "BatchId", "AccountId", "SubscriptionId",
        "CustomerId", "CurrencyCode", "CountryCode", "TransactionStartTime"
    ]
    df_f = df_f.drop(columns=[c for c in drop_cols if c in df_f.columns])
    
    # 6. One-hot encode low-cardinality categorical columns
    cat_cols = ["ProviderId", "ProductId", "ProductCategory", "ChannelId", "PricingStrategy"]
    df_f = pd.get_dummies(df_f, columns=[c for c in cat_cols if c in df_f.columns], drop_first=True, dtype=int)
    
    return df_f

df_processed = engineer_fraud_features(df)
print(f"Processed feature matrix shape: {df_processed.shape}")
```

```python
# Out-of-Time split: train on earliest 80% of transactions, test on latest 20%
split_idx = int(len(df_processed) * 0.80)

train_df = df_processed.iloc[:split_idx]
test_df = df_processed.iloc[split_idx:]

X_train = train_df.drop(columns=["FraudResult"])
y_train = train_df["FraudResult"]

X_test = test_df.drop(columns=["FraudResult"])
y_test = test_df["FraudResult"]

print(f"Train rows: {len(X_train):,} | Fraud cases: {y_train.sum()} ({y_train.mean()*100:.3f}%)")
print(f"Test rows:  {len(X_test):,} | Fraud cases: {y_test.sum()} ({y_test.mean()*100:.3f}%)")
```

```python
from lightgbm import LGBMClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    average_precision_score,
    roc_auc_score,
    precision_recall_curve,
    classification_report,
    confusion_matrix
)

le = LabelEncoder()
X_train["DayOfWeek"] = le.fit_transform(X_train["DayOfWeek"])
X_test["DayOfWeek"]  = le.transform(X_test["DayOfWeek"])
# Calculate imbalance ratio for scale_pos_weight
imbalance_ratio = (y_train == 0).sum() / max((y_train == 1).sum(), 1)
print(f"Class Imbalance Ratio (Legit : Fraud) = {imbalance_ratio:.1f} : 1")

fraud_model = LGBMClassifier(
    n_estimators=250,
    learning_rate=0.03,
    max_depth=5,
    num_leaves=15,
    min_child_samples=50,
    scale_pos_weight=15,  # Forces model to penalize missed fraud heavily
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    verbose=-1
)

fraud_model.fit(X_train, y_train)
fraud_probs = fraud_model.predict_proba(X_test)[:, 1]

pr_auc = average_precision_score(y_test, fraud_probs)
roc_auc = roc_auc_score(y_test, fraud_probs)

print(f"\nOut-of-Time Test PR-AUC  (Key Fraud Metric): {pr_auc:.4f}")
print(f"Out-of-Time Test ROC-AUC:                    {roc_auc:.4f}")
```

```python
# 1. Plot Precision-Recall Curve
precision, recall, pr_thresholds = precision_recall_curve(y_test, fraud_probs)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(recall, precision, color="crimson", linewidth=2.5, label=f"LightGBM (PR-AUC = {pr_auc:.3f})")
axes[0].set_title("Precision-Recall Curve (Out-of-Time Test Set)")
axes[0].set_xlabel("Recall (% of Actual Fraud Caught)")
axes[0].set_ylabel("Precision (% of Alerts That Are True Fraud)")
axes[0].legend()

# 2. UGX Financial Threshold Optimization
AVG_FRAUD_LOSS_UGX = 650_000   # Average UGX loss per missed fraud
ALERT_REVIEW_COST_UGX = 2_000  # Cost of SMS OTP / analyst review per flagged tx

thresholds = np.linspace(0.05, 0.95, 91)
net_ugx_saved = []

for t in thresholds:
    preds = (fraud_probs >= t).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_test, preds).ravel()
    
    prevented_loss = tp * AVG_FRAUD_LOSS_UGX
    review_cost = (tp + fp) * ALERT_REVIEW_COST_UGX
    net_ugx_saved.append((prevented_loss - review_cost) / 1e6)  # In Millions of UGX

best_idx = np.argmax(net_ugx_saved)
best_thresh = thresholds[best_idx]
best_millions = net_ugx_saved[best_idx]

axes[1].plot(thresholds, net_ugx_saved, color="#1f77b4", linewidth=2.5)
axes[1].axvline(best_thresh, color="crimson", linestyle="--", label=f"Optimal Cutoff = {best_thresh:.2f} (UGX {best_millions:.2f}M saved)")
axes[1].set_title("Net Fraud Savings (Million UGX) by Decision Threshold")
axes[1].set_xlabel("Fraud Probability Threshold")
axes[1].set_ylabel("Net Savings (Million UGX)")
axes[1].legend()

plt.tight_layout()
plt.savefig("../models/ugx_fraud_pr_and_roi.png", dpi=150)
plt.show()

# Print final classification report at optimal threshold
opt_preds = (fraud_probs >= best_thresh).astype(int)
print(f"\n--- Classification Report at Optimal Threshold ({best_thresh:.2f}) ---")
print(classification_report(y_test, opt_preds, target_names=["Legitimate (0)", "Fraud (1)"]))
```

```python
import shap
import joblib

# Sample up to 3,000 test rows for fast SHAP computation on CPU
X_sample = X_test.sample(n=min(3000, len(X_test)), random_state=42)

explainer = shap.TreeExplainer(fraud_model)
shap_values = explainer.shap_values(X_sample)
shap_fraud = shap_values[1] if isinstance(shap_values, list) else shap_values

plt.figure(figsize=(10, 6))
shap.summary_plot(shap_fraud, X_sample, show=False)
plt.title("SHAP Feature Drivers — Ugandan Digital Transaction Fraud")
plt.tight_layout()
plt.savefig("../models/fraud_shap_summary.png", dpi=150, bbox_inches="tight")
plt.show()

# Save model + optimal threshold + feature schema
artifact = {
    "model": fraud_model,
    "feature_names": list(X_train.columns),
    "optimal_threshold": float(best_thresh)
}
joblib.dump(artifact, "../models/uganda_fraud_lgbm_v1.joblib")
print(f"Saved model artifact with optimal threshold {best_thresh:.2f}!")
```
